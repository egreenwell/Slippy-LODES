```python
from pygris import tracts
from pygris.data import get_lodes
from pyproj import CRS
import matplotlib.pyplot as plt
import pandas as pd
import geopandas as gpd
import geoplot as gplt
import geoplot.crs as gcrs
import mapclassify as mc
from shapely.geometry import Point
from shapely.geometry import MultiPoint
import folium
import geojson
import requests
```

```python
#step 1 get lodes
va_lodes_od = get_lodes(
    state = "VA",
    year = 2021,
    lodes_type = "od")

va_lodes_od.head()
```

```python
#checking counts of those living outside region and those working outside our region 
df = (
    va_lodes_od
    .assign(w_tract = va_lodes_od['w_geocode'].str.slice(stop = 11),
            h_tract = va_lodes_od['h_geocode'].str.slice(stop = 11))
    .query('w_tract.str.startswith("517600")')
    .groupby(['w_tract', 'h_tract'], as_index = False)
    .agg({'S000': sum})
    .rename({'S000': 'workers'}, axis = 1)
)
```

```python
#step 2 assign a census tract to be a work destination
#and assign a count from each home tract that goes to that destination
#vcu main campus tract 	51760040300
#another to test would be the Cenus tract for MCV/Biotech Park 51760030200
vcu = (
    va_lodes_od
    .assign(w_tract = va_lodes_od['w_geocode'].str.slice(stop = 11),
            h_tract = va_lodes_od['h_geocode'].str.slice(stop = 11))
    .query('w_tract == "51760040300"')
    .groupby('h_tract', as_index = False)
    .agg({'S000': sum})
    .rename({'S000': 'vcu'}, axis = 1)
)

vcu.head()
```

```python
#make a rate for that commute desintation
vcu_commuters = (
    va_lodes_od
    .assign(h_tract = va_lodes_od['h_geocode'].str.slice(stop = 11))
    .groupby('h_tract', as_index = False)
    .agg({'S000': sum})
    .rename({'S000': 'total_workers'}, axis = 1)
    .merge(vcu, on = 'h_tract')
    .assign(vcu_per_1000 = lambda x: 1000 * (x['vcu'] / x['total_workers']))
)

vcu_commuters.head()
```

```python
#saving to excel - just in case - 
vcu_commuters.to_excel(r'C:\GreenGIS\Sankey\vcu.xls', index=False)
```

```python
#find our region tracts - plot to ensure it looks right 
planrva_tracts = tracts(state = "VA", cb = True,
                    county = ["Charles City", "Chesterfield", "Goochland", "Henrico",
                              "Hanover", "New Kent", "Powhatan", "Richmond City"], 
                    year = 2021)
             
planrva_tracts.plot()
planrva_tracts.head()
```

```python
#join and make static figure

vcu_region = planrva_tracts.merge(vcu_commuters, left_on = "GEOID", right_on = "h_tract",
                             how = "left")

vcu_region.fillna(0, inplace = True)

vcu_region.plot(column = 'vcu_per_1000', legend = True, 
               cmap = "cividis", figsize = (8, 8), 
               k = 7, scheme = "naturalbreaks",
               legend_kwds = {"loc": "lower left"})

plt.title("Population Commuting to VCU Main Campus\n(rate per 1000 total commuters)\n2021 LODES data, PlanRVA Census tracts", fontsize = 12)

ax = plt.gca()

ax.set_axis_off()

plt.savefig(r'C:\GreenGIS\Sankey\vcu2.png')
```

```python
#join and make slippy feature

from pygris.geocode import geocode


vcu_region_sub = vcu_region.filter(['GEOID', 'total_workers',
                                  'vcu', 'vcu_per_1000', 'geometry'])

monroe_park = geocode("620 W Main St, Richmond, VA 23220",
                         as_gdf = True)

m = vcu_region_sub.explore(column = "vcu_per_1000", cmap = "cividis",
                          k = 7, scheme = "naturalbreaks", popup = True, 
                          tooltip = False,                   
                          tiles = "CartoDB positron", 
                          style_kwds = {"weight": 0.5},
                          legend_kwds = {"caption": "Population Commuting to VCU Main Campus",
                                          "colorbar": False},
                          popup_kwds = {"aliases": ['Census tract', 'Total workers',
                                                    'VCU-area commuters', 'Rate per 1000']})

monroe_park.explore(m = m, marker_type = "marker", tooltip = False)  
```

```python
#step 2 assign a census tract to be a work destination
#and assign a count from each home tract that goes to that destination
#capital district tract 51760040300
#another to test would be the Cenus tract for MCV/Biotech Park 51760030200
cap = (
    va_lodes_od
    .assign(w_tract = va_lodes_od['w_geocode'].str.slice(stop = 11),
            h_tract = va_lodes_od['h_geocode'].str.slice(stop = 11))
    .query('w_tract == "51760030502"')
    .groupby('h_tract', as_index = False)
    .agg({'S000': sum})
    .rename({'S000': 'cap'}, axis = 1)
)

cap.head()
```

<h1>Using deck.gl to map job locations</h1>

```python
from pygris.data import get_lodes
import pydeck
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

```python
#step 1 get lodes
va_lodes_od = get_lodes(
    state = "VA",
    year = 2021,
    lodes_type = "od",
    agg_level = "tract", 
    return_lonlat = True)

va_lodes_od.head()
```

```python
top_commutes = va_lodes_od.query('w_geocode == "51036" & S000 >= 25')
```

```python
to_charles_city = va_lodes_od.query('w_geocode.str.startswith("510366") & S000 >= 25')
to_charles_city.head()
```

```python
import pydeck

GREEN_RGB = [0, 255, 0, 200]
RED_RGB = [240, 100, 0, 200]

arc_layer = pydeck.Layer(
  "ArcLayer",
  data=top_commutes,
  get_width="S000 / 5",
  get_source_position=["h_lon", "h_lat"],
  get_target_position=["w_lon", "w_lat"],
  get_tilt=15,
  get_source_color=RED_RGB,
  get_target_color=GREEN_RGB,
  pickable=True,
  auto_highlight=True
)

view_state = pydeck.ViewState(
  latitude=37.708664,
  longitude=-77.360546, 
  bearing=0, 
  pitch=50, 
  zoom=8
)

tooltip = {"html": "{S000} jobs <br /> Home of commuter in red; work location in green"}
r = pydeck.Deck(
  arc_layer, 
  initial_view_state=view_state, 
  tooltip=tooltip, 
  map_style = "road"
)

r.to_html("top_commutes.html")
```

```python
#step 1 get lodes
va_lodes_wac = get_lodes(
    state = "VA",
    year = 2021,
    lodes_type = "wac",
    agg_level = "tract", 
    return_lonlat = True)

va_lodes_od.head()
```

```python
def column_to_rgba(column, cmap, alpha):
    normalized = (column - column.min()) / (column.max() - column.min())
    my_cmap = plt.get_cmap(cmap)
    colors = normalized.apply(lambda x: [int(i * 255) for i in mcolors.to_rgba(my_cmap(x, alpha = alpha))])

    return colors
  
  
va_lodes_wac['color'] = column_to_rgba(va_lodes_wac['CNS18'], "viridis", 0.6)
```

```python
layer = pydeck.Layer(
  "ColumnLayer",
  va_lodes_wac,
  get_position=["w_lon", "w_lat"],
  auto_highlight=True,
  elevation_scale=20,
  pickable=True,
  get_elevation = "CNS18",
  get_fill_color = "color",
  elevation_range=[0, 1000],
  extruded=True,
  coverage=1
)

view_state = pydeck.ViewState(
  latitude=37.708664,
  longitude=-77.360546, 
  bearing=0, 
  pitch=50, 
  zoom=6,
  min_zoom=5,
  max_zoom=15,
)

tooltip = {"html": "Number of accommodation / food service jobs: {CNS18}"}

# Render
r = pydeck.Deck(
  layers=[layer], 
  initial_view_state=view_state, 
  map_style = "light", 
  tooltip = tooltip
)

r.to_html("va_service.html")
```
